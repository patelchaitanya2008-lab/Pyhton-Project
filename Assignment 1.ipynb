```python
# Create dictionary and assign 5 elements
my_dict = {
    "name": "ABC",
    "rollno": 34,
    "panel": "B",
    "age": 20,
    "city": "Mumbai"
}

# Display all student information
print("Student information:")
print(my_dict)

# Add new student information
my_dict["email"] = "abc@example.com"
my_dict["phone"] = "9876543210"

print("\nAfter adding email and phone:")
print(my_dict)

# Delete some student information
del my_dict["city"]          # delete using del
my_dict.pop("age", None)     # delete using pop (safe if key missing)

print("\nAfter deleting city and age:")
print(my_dict)

# Display specific student information
print("\nName:", my_dict["name"])
print("Roll No:", my_dict["rollno"])
print("Panel:", my_dict["panel"])
```

```python
my_dict = {
    'name': 'ABC',
    'panel': 'B',
    'rollno': 34,
    'marks': [65, 87, 67, 94]
}

# sorted(my_dict) returns sorted list of keys
sorted_keys = sorted(my_dict)
print("Sorted keys:", sorted_keys)

# If you want a new dict with keys sorted:
sorted_dict = {k: my_dict[k] for k in sorted_keys}
print("Sorted dictionary by keys:", sorted_dict)
```

```python
my_dict = {
    'name': 'ABC',
    'panel': 'B',
    'rollno': 34,
    'marks': [65, 87, 67, 94]
}

keys_list = list(my_dict.keys())
values_list = list(my_dict.values())

print("Keys list:", keys_list)
print("Values list:", values_list)
```

```python
my_dict = {
    'marks1': 23,
    'marks2': 123,
    'marks3': 43,
    'marks4': 13,
    'marks5': 39
}

values = list(my_dict.values())
mean_value = sum(values) / len(values)

print("Mean of all values:", mean_value)
```

```python
my_dict = {
    'name': ['Yash', 'Neel', 'Dev'],
    'rollno': [11, 12, 13],
    'marks': [78, 56, 98]
}

# a) Display name as Dev
# Assuming you want the element 'Dev' from the name list
print("Name as Dev:", my_dict['name'][2])  # index 2

# b) Display roll no 12
# Find index where rollno is 12, then show it
idx = my_dict['rollno'].index(12)
print("Roll no 12:", my_dict['rollno'][idx])

# c) Display greatest marks
greatest_marks = max(my_dict['marks'])
print("Greatest marks:", greatest_marks)
```

```python
s = input("Enter a string: ")

freq = {}
for ch in s:
    if ch.isalpha():  # count only letters
        ch_lower = ch.lower()
        freq[ch_lower] = freq.get(ch_lower, 0) + 1

print("Frequency of letters:", freq)
```

```python
numerator = int(input("Enter numerator: "))
denominator = int(input("Enter denominator: "))

quotient = numerator // denominator
remainder = numerator % denominator

result = (quotient, remainder)
print("Quotient and Remainder:", result)
```
