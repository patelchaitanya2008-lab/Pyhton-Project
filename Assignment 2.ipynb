```python
# Accept marks in Math
marks = float(input("Enter marks in Math (0–100): "))

# Validate marks (optional but good practice)
if marks < 0 or marks > 100:
    print("Invalid marks. Please enter a value between 0 and 100.")
else:
    # Determine grade using if...elif...else
    if marks >= 90:          # 90–100
        grade = 'O'
    elif marks >= 80:        # 80–89
        grade = 'A+'
    elif marks >= 70:        # 70–79
        grade = 'A'
    elif marks >= 60:        # 60–69
        grade = 'B'
    elif marks >= 50:        # 50–59
        grade = 'C'
    elif marks >= 40:        # 40–49
        grade = 'P'
    else:                    # Below 40
        grade = 'F'

    print("Grade obtained:", grade)
```

```python
x = 10
y = 20

# Check if x is positive AND y is positive
result = (x > 0) and (y > 0)
print(result)  # True, because both are > 0

result = (x > 0) and (y < 0)
print(result)  # False, because y < 0 is False
```

```python
x = 10
y = -5

# Check if x is positive OR y is positive
result = (x > 0) or (y > 0)
print(result)  # True, because x > 0

result = (x < 0) or (y < 0)
print(result)  # True, because y < 0

result = (x < 0) or (y > 0)
print(result)  # False, both are False
```

```python
x = 10

result = not (x > 5)
print(result)  # False, because (x > 5) is True, so not True → False

result = not (x < 5)
print(result)  # True, because (x < 5) is False, so not False → True
```
