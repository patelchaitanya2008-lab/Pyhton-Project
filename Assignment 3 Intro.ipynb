```python
import math


def find_roots(a, b, c):
    try:
        # float(None) raises TypeError.
        a = float(a)
        b = float(b)
        c = float(c)

        if a == 0:
            raise ValueError("a must not be zero for a quadratic equation.")

        discriminant = b**2 - 4*a*c

        # A negative discriminant raises ValueError: math domain error.
        sqrt_d = math.sqrt(discriminant)

        root1 = (-b + sqrt_d) / (2*a)
        root2 = (-b - sqrt_d) / (2*a)

        if discriminant == 0:
            print("The roots are real and equal.")
        else:
            print("The roots are real and different.")

        print("Root 1:", root1)
        print("Root 2:", root2)

    except ValueError as error:
        print("ValueError:", error)

    except TypeError as error:
        print("TypeError: Coefficients must be numbers, not None "
              "or other unsupported types.")
        print("Details:", error)


a = input("Enter coefficient a: ")
b = input("Enter coefficient b: ")
c = input("Enter coefficient c: ")

find_roots(a, b, c)

# To test handling of None, uncomment:
# find_roots(None, 2, 3
```
